```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("IRIS.csv")

# Seaborn plots
sns.histplot(df["petal-length"])
sns.histplot(df["sepal-length"])
plt.title("Petal Length and Sepal Length Histogram")
plt.show()

sns.boxplot(data=df[["sepal-length", "sepal-width", "petal-length", "petal-width"]])
plt.title("Box Plot of All Iris Features")
plt.show()

sns.scatterplot(data=df, x="sepal-length", y="petal-length", hue="species")
plt.title("Sepal Length and Petal Length by Species")
plt.show()

sns.pairplot(df, hue="species")
plt.show()

correlation = df[["sepal-length", "sepal-width", "petal-length", "petal-width"]].corr()
sns.heatmap(correlation, annot=True)
plt.title("Correlation of Iris Features")
plt.show()


plt.subplot(1, 2, 1)
plt.hist(df["petal-length"], bins=10)
plt.title("Petal Length")

plt.subplot(1, 2, 2)
plt.hist(df["sepal-length"], bins=10)
plt.title("Sepal Length")
plt.show()

plt.boxplot([df["sepal-length"], df["sepal-width"], df["petal-length"], df["petal-width"]], tick_labels=["Sepal Length", "Sepal Width", "Petal Length", "Petal Width"])
plt.title("Box Plot of All Iris Features")
plt.show()

plt.scatter(df["sepal-length"], df["petal-length"])
plt.title("Sepal Length and Petal Length")
plt.show()

pd.plotting.scatter_matrix(df[["sepal-length", "sepal-width", "petal-length", "petal-width"]])
plt.show()

plt.matshow(correlation)
plt.xticks(range(4), correlation.columns)
plt.yticks(range(4), correlation.columns)
plt.title("Correlation of Iris Features")
plt.show()
```
